# Action Recommendation Engine

```python
def assign_risk(prob):
    if prob < 0.3:
        return "Low Risk"
    elif prob < 0.6:
        return "Medium Risk"
    else:
        return "High Risk"
```

```python
def recommend_action(risk):
    if risk == "Low Risk":
        return "Send payment reminder via SMS/Email"
    elif risk == "Medium Risk":
        return "Offer flexible EMI or short-term payment plan"
    else:
        return "Assign to recovery agent and initiate call"
```

```python
def generate_recommendation(prob):
    risk = assign_risk(prob)
    action = recommend_action(risk)
    return {
        "risk_level": risk,
        "recommended_action": action
    }
```

# FastAPI Risk Scoring

```python
from fastapi import FastAPI
from pydantic import BaseModel
import joblib
import numpy as np

app = FastAPI(title="CreditPathAI Risk Scoring API")

model = joblib.load("best_model.pkl")

class InputData(BaseModel):
    features: list[float]

def assign_risk(prob):
    if prob < 0.3:
        return "Low Risk"
    elif prob < 0.6:
        return "Medium Risk"
    else:
        return "High Risk"

def recommend_action(risk):
    if risk == "Low Risk":
        return "Send payment reminder via SMS/Email"
    elif risk == "Medium Risk":
        return "Offer flexible EMI or short-term payment plan"
    else:
        return "Assign to recovery agent and initiate call"

@app.post("/predict")
def predict(data: InputData):
    features_array = np.array(data.features).reshape(1, -1)
    prob = model.predict_proba(features_array)[0][1]

    risk = assign_risk(prob)
    action = recommend_action(risk)

    return {
        "default_probability": round(float(prob), 3),
        "risk_level": risk,
        "recommended_action": action
    }
```
