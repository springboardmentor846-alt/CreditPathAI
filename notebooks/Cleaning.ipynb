# Cleaning

```python
import pandas as pd 
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report
```

```python
DATA_PATH = "../data/raw/Loan_Default.csv"   # change path if needed
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df=df.drop_duplicates()

num_cols=df.select_dtypes(include=['int64','float64']).columns
df[num_cols]=df[num_cols].fillna(df[num_cols].median())

cat_cols=df.select_dtypes(include=['object']).columns
df[cat_cols]=df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
df.columns
```

```python
x=df.drop("Default", axis=1)
y=df["Default"]
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.dtypes)
```

```python
x_train = x_train.drop(columns=['LoanID'])
x_test  = x_test.drop(columns=['LoanID'])
```

```python
cat_cols = x_train.select_dtypes(include='object').columns

x_train = pd.get_dummies(x_train, columns=cat_cols, drop_first=True)
x_test  = pd.get_dummies(x_test, columns=cat_cols, drop_first=True)

# align train & test columns
x_train, x_test = x_train.align(x_test, join='left', axis=1, fill_value=0)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test  = scaler.transform(x_test)
```

```python
log_reg = LogisticRegression(max_iter=1000,class_weight='balanced')
log_reg.fit(x_train, y_train)
```

```python
y_pred = log_reg.predict(x_test)
y_pred_prob = log_reg.predict_proba(x_test)[:,1]

y_pred_custom=(y_pred_prob >= 0.3).astype(int)
auc = roc_auc_score(y_test, y_pred_prob)
print("AUC-ROC Score:", auc)
```

```python
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test,y_pred_custom))
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv("Loan_Default_Cleaned.csv")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.columns
```

```python
if 'ID' in df.columns:
    df.drop(['ID'], axis=1, inplace=True)
```

```python
for col in df.select_dtypes(include='object'):
    df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
df['DTIRatio'] = df['DTIRatio'].fillna(df['DTIRatio'].median())
```

```python
df['CreditScore'] = df['CreditScore'].fillna(df['CreditScore'].quantile(0.25))
```

```python
for col in df.select_dtypes(include=np.number):
    df[col] = df[col].fillna(df[col].median())
```

```python
for col in df.select_dtypes(include='object'):
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
if df['Default'].isna().sum() > 0:
    df['Default'] = df['Default'].fillna('very_poor')
```

```python
if 'Age' in df.columns:

    age_map = {
        '18-24': 1, '25-34': 2, '35-44': 3,
        '45-54': 4, '55-64': 5, '65-74': 6,
        '65+': 6, '75+': 6, '70+': 6
    }

    df['age_group'] = df['Age'].map(age_map)

    if df['age_group'].dropna().shape[0] > 0:
        df['age_group'] = df['age_group'].fillna(df['age_group'].median())
    else:
        df['age_group'] = df['age_group'].fillna(3)  
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_cols = df.select_dtypes(include='object').columns
le = LabelEncoder()
```

```python
for col in label_cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
num_cols = df.select_dtypes(include=np.number).columns.drop(['Default']) 
scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
df.to_csv("Loan_Default_Preprocessed.csv", index=False)
```
